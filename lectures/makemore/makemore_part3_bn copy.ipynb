````python
# build the vocabulary of characters and mappings to/from integers
chars = sorted(list(set(''.join(words))))
stoi = {s:i+1 for i,s in enumerate(chars)}
stoi['.'] = 0
itos = {i:s for s,i in stoi.items()}
vocab_size = len(itos)
print(itos)
print(vocab_size)
````

````python
# build the dataset
block_size = 3 # context length: how many characters do we take to predict the next one?

def build_dataset(words):
  X, Y = [], []

  for w in words:
    context = [0] * block_size
    for ch in w + '.':
      ix = stoi[ch]
      X.append(context)
      Y.append(ix)
      context = context[1:] + [ix] # crop and append

  X = torch.tensor(X)
  Y = torch.tensor(Y)
  print(X.shape, Y.shape)
  return X, Y

import random
random.seed(42)
random.shuffle(words)
n1 = int(0.8*len(words))
n2 = int(0.9*len(words))

Xtr,  Ytr  = build_dataset(words[:n1])     # 80%
Xdev, Ydev = build_dataset(words[n1:n2])   # 10%
Xte,  Yte  = build_dataset(words[n2:])     # 10%
````

## Initialization Problems

### First pass

The first pass of random initialization shown gave a loss of 27.8. This is surprisingly high.

We would expect that truly random guesses would have a 1/27 chance of being correct, so our loss function would be $-\ln{\frac{1}{27}}$, which is much smaller:

````python
expected = -torch.asarray(1./27).log()
print(f"expected random loss: {expected}")
````

What this high loss is telling us is that the output isn't random; it's preferring certain wrong outcomes - i.e., it's confidently wrong.

So:

- We set `b2` to zero - these are biases, they can be zero
- We multiply `W2` by 0.01 - we want these to be very small, but initialized by something so they stay independent

Then our loss decreases look like more like a normal decrease, rather than a hockey stick.

### Problems with `h`

At first, our hidden layer has a problem too. Our hidden layer is calculated as $h = \tanh\left(E w + b\right)$, as before; and at this point (well, [this point in the video](https://www.youtube.com/watch?v=P6sfmUTpUmc&t=779s)), most of the values in $h$ are either 1.0 or -1.0. At that extreme, the derivative of $h$ is roughly zero, so it's not very susceptible to optimization.

- So then we can do roughly the same - `b1 = torch.randn(…) * 0.01`, and `W1 = torch.randn(…) * 0.01`.

### Generalizing initialization issues

If you start with two gaussian random arrays and do $y = x w$, then $\mathrm{std}(y) = \mathrm{std}(x) \mathrm{std}(y) \sqrt{d}$, where $d$ is the number of fan-in dimensions.

So to generalize: if we normalize by dividing by $\sqrt{d}$, then the output of that layer should start with a random normal distribution of stddev 1.

Then because we also have a `tanh` function, that squashes it to some degree, so we 'add that back in' - with a gain of $\frac{5}{3}$. 

[Gemini](https://gemini.google.com/app/15f51fed60d3012c) doesn't think there's a closed form solution for this, but the simulated value is about 0.6279, close to the below. And... 5/3 is close enough to 1 over that.

Anyways - what this means is that for `W1`, we multiply by $\frac{5}{3} \sqrt{d}$

````python
x = torch.randn(10000000, 1)
std = x.std()
tanh_std = x.tanh().std()
print(f'std in: {std}; tanh std: {tanh_std}')
````

````python
# MLP revisited
n_embd = 10 # the dimensionality of the character embedding vectors
n_hidden = 200 # the number of neurons in the hidden layer of the MLP

g = torch.Generator().manual_seed(2147483647) # for reproducibility
C  = torch.randn((vocab_size, n_embd),            generator=g)
W1 = torch.randn((n_embd * block_size, n_hidden), generator=g) * (5/3)/((n_embd * block_size)**0.5) #* 0.2
#b1 = torch.randn(n_hidden,                        generator=g) * 0.01
W2 = torch.randn((n_hidden, vocab_size),          generator=g) * 0.01
b2 = torch.randn(vocab_size,                      generator=g) * 0
````

### Batch Normalization

Insight: instead of calculating this from first principles, we can instead... just normalize it to unit gaussian per batch / per neuron.

We do this by forcing it, but then adding an additional set of parameters for the means and stddevs, and we then allow the neural network to modify those as it needs to.

Note that in doing so, we can drop the 'bias' from the batch-normalized layer; it's now included in the batch normalization.

This stabilizes the values - both at initialization, and during training, keeping them from growing so large (positive or negative) that they saturate the $\tanh$ and the derivatives become zero and no more learning happens. But it does mean that the examples are more 'coupled' to each other.

````python
# BatchNorm parameters
bngain = torch.ones((1, n_hidden))
bnbias = torch.zeros((1, n_hidden))
bnmean_running = torch.zeros((1, n_hidden))
bnstd_running = torch.ones((1, n_hidden))

parameters = [C, W1, W2, b2, bngain, bnbias]
print(sum(p.nelement() for p in parameters)) # number of parameters in total
for p in parameters:
  p.requires_grad = True
````

#### Running Normalization

We also keep a 'running' mean and stddev estimation, so we can use it at the end instead of the batch ones, and then run it on single examples.

````python
# same optimization as last time
max_steps = 200000
batch_size = 32
lossi = []

for i in range(max_steps):

  # minibatch construct
  ix = torch.randint(0, Xtr.shape[0], (batch_size,), generator=g)
  Xb, Yb = Xtr[ix], Ytr[ix] # batch X,Y

  # forward pass
  emb = C[Xb] # embed the characters into vectors
  embcat = emb.view(emb.shape[0], -1) # concatenate the vectors
  # Linear layer
  hpreact = embcat @ W1 #+ b1 # hidden layer pre-activation
  # BatchNorm layer
  # -------------------------------------------------------------
  bnmeani = hpreact.mean(0, keepdim=True)
  bnstdi = hpreact.std(0, keepdim=True)
  hpreact = bngain * (hpreact - bnmeani) / bnstdi + bnbias
  with torch.no_grad():
    bnmean_running = 0.999 * bnmean_running + 0.001 * bnmeani
    bnstd_running = 0.999 * bnstd_running + 0.001 * bnstdi
  # -------------------------------------------------------------
  # Non-linearity
  h = torch.tanh(hpreact) # hidden layer
  logits = h @ W2 + b2 # output layer
  loss = F.cross_entropy(logits, Yb) # loss function

  # backward pass
  for p in parameters:
    p.grad = None
  loss.backward()

  # update
  lr = 0.1 if i < 100000 else 0.01 # step learning rate decay
  for p in parameters:
    p.data += -lr * p.grad

  # track stats
  if i % 10000 == 0: # print every once in a while
    print(f'{i:7d}/{max_steps:7d}: {loss.item():.4f}')
  lossi.append(loss.log10().item())
````

````python
# calibrate the batch norm at the end of training

with torch.no_grad():
  # pass the training set through
  emb = C[Xtr]
  embcat = emb.view(emb.shape[0], -1)
  hpreact = embcat @ W1 # + b1
  # measure the mean/std over the entire training set
  bnmean = hpreact.mean(0, keepdim=True)
  bnstd = hpreact.std(0, keepdim=True)
````

````python
@torch.no_grad() # this decorator disables gradient tracking
def split_loss(split):
  x,y = {
    'train': (Xtr, Ytr),
    'val': (Xdev, Ydev),
    'test': (Xte, Yte),
  }[split]
  emb = C[x] # (N, block_size, n_embd)
  embcat = emb.view(emb.shape[0], -1) # concat into (N, block_size * n_embd)
  hpreact = embcat @ W1 # + b1
  #hpreact = bngain * (hpreact - hpreact.mean(0, keepdim=True)) / hpreact.std(0, keepdim=True) + bnbias
  hpreact = bngain * (hpreact - bnmean_running) / bnstd_running + bnbias
  h = torch.tanh(hpreact) # (N, n_hidden)
  logits = h @ W2 + b2 # (N, vocab_size)
  loss = F.cross_entropy(logits, y)
  print(split, loss.item())

split_loss('train')
split_loss('val')
````

### Batch Norm Folding

At the end, we now have the running mean $\mu = \texttt{bnmean\_running}$, the running std dev $\sigma = \texttt{bnstd\_running}$, the gain $\gamma = \texttt{bngain}$, and the bias $\beta = \texttt{bnbias}$. These can be 'folded' into each other. If $Y$ is the output and $X$ the input, we have:

$$
\begin{align}
Y &= \gamma * \left(\frac{X - \mu}{sigma}\right) + \beta \\
Y &= \left(\frac{\gamma}{\sigma}\right) X + \left(\beta - \frac{\gamma \mu}{ \sigma}\right) 
\gamma' &= \frac{\gamma}{\sigma} & \beta' = \beta - \frac{\gamma \mu}{ \sigma}\\
Y &= \gamma' X + \beta'
\end{align} 
$$

Then we can just use a gain of $\gamma'$ and a bias of $\beta'$, and forget about the mean / stddev.

You can even go one step further: fold the gain into the weights themselves.

### Batch Normalization: Conclusions

It came out in 2015, and was very influential, because it controlled these activations and gave a way to normalize them that was still differentiable (with the scaling). However, it adds some significant complexity, and Karpathy has 'shot himself in the foot' a number of times with it, and wants to save us from that

````python
# Let's train a deeper network
# The classes we create here are the same API as nn.Module in PyTorch

class Linear:

  def __init__(self, fan_in, fan_out, bias=True):
    self.weight = torch.randn((fan_in, fan_out), generator=g) / fan_in**0.5
    self.bias = torch.zeros(fan_out) if bias else None

  def __call__(self, x):
    self.out = x @ self.weight
    if self.bias is not None:
      self.out += self.bias
    return self.out

  def parameters(self):
    return [self.weight] + ([] if self.bias is None else [self.bias])


class BatchNorm1d:

  def __init__(self, dim, eps=1e-5, momentum=0.1):
    self.eps = eps
    self.momentum = momentum
    self.training = True
    # parameters (trained with backprop)
    self.gamma = torch.ones(dim)
    self.beta = torch.zeros(dim)
    # buffers (trained with a running 'momentum update')
    self.running_mean = torch.zeros(dim)
    self.running_var = torch.ones(dim)

  def __call__(self, x):
    # calculate the forward pass
    if self.training:
      xmean = x.mean(0, keepdim=True) # batch mean
      xvar = x.var(0, keepdim=True) # batch variance
    else:
      xmean = self.running_mean
      xvar = self.running_var
    xhat = (x - xmean) / torch.sqrt(xvar + self.eps) # normalize to unit variance
    self.out = self.gamma * xhat + self.beta
    # update the buffers
    if self.training:
      with torch.no_grad():
        self.running_mean = (1 - self.momentum) * self.running_mean + self.momentum * xmean
        self.running_var = (1 - self.momentum) * self.running_var + self.momentum * xvar
    return self.out

  def parameters(self):
    return [self.gamma, self.beta]

class Tanh:
  def __call__(self, x):
    self.out = torch.tanh(x)
    return self.out
  def parameters(self):
    return []

class GeLU:
  def __init__(self):
    self._gelu = torch.nn.GELU()

  def __call__(self, x):
    self.out = self._gelu(x)
    return self.out

  def parameters(self):
    return []

n_embd = 10 # the dimensionality of the character embedding vectors
n_hidden = 100 # the number of neurons in the hidden layer of the MLP
g = torch.Generator().manual_seed(2147483647) # for reproducibility

C = torch.randn((vocab_size, n_embd),            generator=g)
# layers = [
#   Linear(n_embd * block_size, n_hidden, bias=False), BatchNorm1d(n_hidden), Tanh(),
#   Linear(           n_hidden, n_hidden, bias=False), BatchNorm1d(n_hidden), Tanh(),
#   Linear(           n_hidden, n_hidden, bias=False), BatchNorm1d(n_hidden), Tanh(),
#   Linear(           n_hidden, n_hidden, bias=False), BatchNorm1d(n_hidden), Tanh(),
#   Linear(           n_hidden, n_hidden, bias=False), BatchNorm1d(n_hidden), Tanh(),
#   Linear(           n_hidden, vocab_size, bias=False), BatchNorm1d(vocab_size),
# ]
# layers = [
#   Linear(n_embd * block_size, n_hidden), Tanh(),
#   Linear(           n_hidden, n_hidden), Tanh(),
#   Linear(           n_hidden, n_hidden), Tanh(),
#   Linear(           n_hidden, n_hidden), Tanh(),
#   Linear(           n_hidden, n_hidden), Tanh(),
#   Linear(           n_hidden, vocab_size),
# ]

layers = [
  Linear(n_embd * block_size, n_hidden, bias=False), BatchNorm1d(n_hidden), GeLU(),
  Linear(           n_hidden, n_hidden, bias=False), BatchNorm1d(n_hidden), GeLU(),
  Linear(           n_hidden, n_hidden, bias=False), BatchNorm1d(n_hidden), GeLU(),
  Linear(           n_hidden, n_hidden, bias=False), BatchNorm1d(n_hidden), GeLU(),
  Linear(           n_hidden, n_hidden, bias=False), BatchNorm1d(n_hidden), GeLU(),
  Linear(           n_hidden, vocab_size, bias=False), BatchNorm1d(vocab_size),
]

with torch.no_grad():
  # last layer: make less confident
  layers[-1].gamma *= 0.1
  #layers[-1].weight *= 0.1
  # all other layers: apply gain
  for layer in layers[:-1]:
    if isinstance(layer, Linear):
      layer.weight *= 1.0 #5/3 # 1 for no tanh / with batch-norm; for just tanh, 5/3 works

parameters = [C] + [p for layer in layers for p in layer.parameters()]
print(sum(p.nelement() for p in parameters)) # number of parameters in total
for p in parameters:
  p.requires_grad = True
````

````python
# same optimization as last time
max_steps = 200000
batch_size = 32
# Loss at each step (NOTE: removed log10)
lossi = []
# Calculate update-to-weight ratios (NOTE: removed log10)
ud = []

steps = []
for i in range(max_steps):

  # minibatch construct
  ix = torch.randint(0, Xtr.shape[0], (batch_size,), generator=g)
  Xb, Yb = Xtr[ix], Ytr[ix] # batch X,Y

  # forward pass
  emb = C[Xb] # embed the characters into vectors
  x = emb.view(emb.shape[0], -1) # concatenate the vectors
  for layer in layers:
    x = layer(x)
  loss = F.cross_entropy(x, Yb) # loss function

  # backward pass
  for layer in layers:
    layer.out.retain_grad() # AFTER_DEBUG: would take out retain_graph
  for p in parameters:
    p.grad = None
  loss.backward()

  # update
  lr = 0.1 if i < 150000 else 0.01 # step learning rate decay
  for p in parameters:
    p.data += -lr * p.grad

  # track stats
  steps.append(lr)
  if i % 10000 == 0: # print every once in a while
    print(f'{i:7d}/{max_steps:7d}: {loss.item():.4f}')
  lossi.append(loss.item())
  with torch.no_grad():
    ud.append([((lr*p.grad).std() / p.data.std()).item() for p in parameters])

  # if i >= 1000:
  #   break # AFTER_DEBUG: would take out obviously to run full optimization
````

````python
import numpy as np

def gaussian_moving_average(x, data, sigma=3):
    radius = int(3 * sigma)
    print(radius)
    xw = np.arange(-radius, radius + 1)
    w = np.exp(-(xw**2) / (2 * sigma**2))
    w /= w.sum()
    print(len(w))

    xs = np.asarray(x)[radius:-radius]

    return xs, np.convolve(data, w, mode="valid")

stepsx = np.cumsum(steps)
plt.plot(stepsx, lossi)
plt.plot(*gaussian_moving_average(stepsx, lossi, len(stepsx)//200), color='black')

plt.yscale('log')
````
